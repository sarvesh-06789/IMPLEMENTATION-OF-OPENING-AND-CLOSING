```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. Create a blank black image
image = np.zeros((600, 600, 3), dtype=np.uint8)
```

```python
# 2. Add text on the image
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(image, 'Open and Close', (100, 250), font, 2, (255, 255, 255), 2, cv2.LINE_AA)
```

```python
# 3. Create a 3x3 structuring element (kernel)
kernel = np.ones((3, 3), np.uint8)
```

```python
# 4. Perform Opening Operation (Erosion followed by Dilation)
opened_image = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel)
```

```python
# 5. Perform Closing Operation (Dilation followed by Erosion)
closed_image = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel)
```

```python
# 6. Display Results
plt.figure(figsize=(12, 4))
```

```python
# Input Image
plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Input Image with Text")
plt.axis("off")
```

```python
# Opening Operation
plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(opened_image, cv2.COLOR_BGR2RGB))
plt.title("Opening Operation")
plt.axis("off")
```

```python
# Closing Operation
plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(closed_image, cv2.COLOR_BGR2RGB))
plt.title("Closing Operation")
plt.axis("off")
```

```python
plt.tight_layout()
plt.show()
```
